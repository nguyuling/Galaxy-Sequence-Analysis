### **Connect to Galaxy API**

```python
from bioblend.galaxy import GalaxyInstance
import os
from dotenv import load_dotenv

# connect to the galaxy server
load_dotenv()
galaxy_url = "https://usegalaxy.org"
api_key = os.getenv("API_KEY")
gi = GalaxyInstance(url=galaxy_url, key=api_key)

# create a new history for the project
new_history = gi.histories.create_history(name="Kraken2")

# search for a specific metagenomics tool
tools = gi.tools.get_tools(name="Kraken2")
if tools:
    print(f"Found tool: {tools[0]['name']} (ID: {tools[0]['id']})")
```

```python
import os
from bioblend.galaxy import GalaxyInstance
from dotenv import load_dotenv

# connect to the galaxy server (api key store .env file)
load_dotenv()
galaxy_url = "https://usegalaxy.org"
api_key = os.getenv("API_KEY")
gi = GalaxyInstance(url=galaxy_url, key=api_key)

# create history and save its id globally for the notebook
history_name = "Metagenomics_Analysis"
history = gi.histories.create_history(name="history_name")
history_id = history["id"]
print(f"Connected. Active History: '{history_name}' (ID: {history_id})")

# helper function to find tool ids quickly
def get_tool_id(tool_name):
    tools = gi.tools.get_tools(name=tool_name)
    if tools:
        print(f"✓ Found '{tools[0]['name']}': {tools[0]['id']}")
        return tools[0]["id"]
    print(f"✗ Tool '{tool_name}' not found.")
    return None
```
